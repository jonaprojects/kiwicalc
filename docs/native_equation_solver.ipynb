# KiwiCalc native equation solver
This notebook demonstrates exact symbolic answers, periodic families, structured derivations, bounded numerical fallback, and exact systems.

```python
import math
import kiwicalc as kw
```

```python
examples = [
    '2*x + 1 = 5',
    '(x - 2)^2 = 0',
    '(x^2 - 1)/(x - 1) = 0',
    'sqrt(x + 1) = x - 1',
    'abs(2*x - 1) = 3',
    '2^x = 8',
    'ln(x) = 2',
]
for equation in examples:
    result = kw.solve_equation(equation, steps=True)
    print(equation, '->', result.solution_set, 'conditions:', result.conditions)
```

```python
periodic = kw.solve_equation('sin(x) = 0')
bounded = kw.solve_equation('cos(2*x) = 1/2', interval=(-math.pi, math.pi))
print(periodic.solution_set)
print(bounded.solution_set)
```

```python
approximate = kw.solve_equation('cos(x) = x', interval=(0, 1), steps=True)
print(approximate.solutions)
print('residuals:', approximate.residuals, 'evaluations:', approximate.evaluations)
```

```python
unique = kw.solve_equation_system(('x+y=3', 'x-y=1'))
parametric = kw.solve_equation_system(('x+y=2',))
print(unique.solutions)
print(parametric.solutions, parametric.parameters)
```
