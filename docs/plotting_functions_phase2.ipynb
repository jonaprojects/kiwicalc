# Plotting API expansion — Phase 2

Common mathematical constructions: piecewise functions, regions, inequalities, direct parametric and polar plots, sequences, and error bands.

```python
import matplotlib.pyplot as plt
import numpy as np
import kiwicalc as kw
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 9))
piecewise = kw.plot_piecewise(
    [((-3, 0), lambda x: -x, '[)'), ((0, 3), lambda x: x*x/3, '[]')],
    samples=81, title='Piecewise endpoints', fig=fig, ax=axes[0, 0], show=False,
)
region = kw.plot_region(
    lambda x: x*x, lambda x: 2*x+3, interval=(-1, 3),
    color='cornflowerblue', title='Region between curves',
    fig=fig, ax=axes[0, 1], show=False,
)
inequality = kw.plot_inequality(
    'x^2 + y^2 <= 4', x_range=(-3, 3), y_range=(-3, 3),
    samples=100, color='gold', title='x² + y² ≤ 4',
    fig=fig, ax=axes[1, 0], show=False,
)
polar = kw.plot_polar(
    lambda theta: 1+np.cos(theta), samples=300, title='Cardioid',
    fig=fig, ax=axes[1, 1], show=False,
)
fig.tight_layout()
assert len(piecewise.endpoint_markers) == 4
assert inequality.kiwicalc_mask.any()
fig
```

```python
fig2, axes2 = plt.subplots(1, 3, figsize=(14, 4))
parametric = kw.plot_parametric(
    lambda t: np.cos(3*t), lambda t: np.sin(2*t), samples=400,
    title='Parametric', fig=fig2, ax=axes2[0], show=False,
)
sequence = kw.plot_sequence(
    lambda n: (-1)**n/n, start=1, stop=16, title='Sequence',
    fig=fig2, ax=axes2[1], show=False,
)
band = kw.plot_error_band(
    np.sin, lambda x: 0.08+0.02*np.abs(x), start=-np.pi, stop=np.pi,
    title='Error band', fig=fig2, ax=axes2[2], show=False,
)
fig2.tight_layout()
assert len(sequence.kiwicalc_values) == 16
assert np.all(band.kiwicalc_error >= 0)
fig2
```
