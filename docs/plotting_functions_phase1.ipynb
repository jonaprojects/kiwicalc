# Plotting API expansion — Phase 1

Standalone scientific-field functions. Every method accepts supplied Matplotlib axes, returns its primary artist, and keeps related artists and colorbars accessible.

```python
import matplotlib.pyplot as plt
import numpy as np
import kiwicalc as kw
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 9))
rotation = kw.plot_vector_field(
    lambda x, y: -y, lambda x, y: x, density=15,
    x_range=(-3, 3), y_range=(-3, 3), color='magnitude',
    title='Vector field', fig=fig, ax=axes[0, 0], show=False,
)
slope = kw.plot_slope_field(
    lambda x, y: x-y, density=15, title='Slope field',
    fig=fig, ax=axes[0, 1], show=False,
)
stream = kw.plot_streamlines(
    lambda x, y: -y, lambda x, y: x, samples=35, color='teal',
    title='Streamlines', fig=fig, ax=axes[1, 0], show=False,
)
contour = kw.plot_contour(
    lambda x, y: x*x+y*y, levels=[1, 2, 4, 6], labels=True,
    samples=60, title='Contour map', fig=fig, ax=axes[1, 1], show=False,
)
fig.tight_layout()
assert rotation.U.size == 225
assert len(stream.kiwicalc_artists) == 2
assert len(contour.levels) == 4
fig
```

```python
fig2, ax2 = plt.subplots(figsize=(6, 5))
gradient = kw.plot_gradient_field(
    'f(x,y)=x^2-y^2', density=17, color='magnitude', colorbar=True,
    title='Gradient field', fig=fig2, ax=ax2, show=False,
)
assert gradient.kiwicalc_colorbar is not None
fig2
```
